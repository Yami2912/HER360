```python
# Select relevant columns
skills_clean = skills_df[
    ["O*NET-SOC Code", "Title", "Element Name", "Scale ID", "Data Value"]
].copy()

# Convert IM and LV records into separate columns
skills_pivot = skills_clean.pivot(
    index=["O*NET-SOC Code", "Title", "Element Name"],
    columns="Scale ID",
    values="Data Value"
).reset_index()

# Rename columns
skills_pivot = skills_pivot.rename(columns={
    "IM": "Importance",
    "LV": "Level"
})

# Display the processed dataset
display(skills_pivot.head(10))

print("Processed shape:", skills_pivot.shape)
print("\nMissing values:")
print(skills_pivot.isnull().sum())
```

```python
# Load the occupation dataset
occupation_df = datasets["occupation_data.csv"]

# Merge processed skills with occupation descriptions
career_dataset = skills_pivot.merge(
    occupation_df[["O*NET-SOC Code", "Description"]],
    on="O*NET-SOC Code",
    how="left"
)

print("Merged dataset shape:", career_dataset.shape)

display(career_dataset.head(10))

print("\nMissing values:")
print(career_dataset.isnull().sum())
```

```python
import pandas as pd

# 1. Copy raw DataFrames so the originals in skill_files stay untouched
ess_copy = skill_files["essential_skills.csv"].copy()
ess_copy["Skill_Category"] = "Essential"

trans_copy = skill_files["transferable_skills.csv"].copy()
trans_copy["Skill_Category"] = "Transferable"

# 2. Stack them vertically into one raw combined skills table
all_skills_raw = pd.concat([ess_copy, trans_copy], ignore_index=True)
print("Combined raw shape (should be 63,700 rows):", all_skills_raw.shape)

# 3. Pivot IM (Importance) and LV (Level) into separate columns
skills_35_pivot = all_skills_raw.pivot(
    index=["O*NET-SOC Code", "Title", "Element ID", "Element Name", "Skill_Category"],
    columns="Scale ID",
    values="Data Value"
).reset_index()

skills_35_pivot = skills_35_pivot.rename(columns={"IM": "Importance", "LV": "Level"})

# 4. Also bring in the quality flags (Not Relevant from LV rows; Recommend Suppress if Y on either IM or LV)
flags = (
    all_skills_raw.groupby(["O*NET-SOC Code", "Element ID"], as_index=False)
    .agg(
        Recommend_Suppress=("Recommend Suppress", lambda s: "Y" if (s == "Y").any() else "N"),
        Not_Relevant=("Not Relevant", lambda s: "Y" if (s == "Y").any() else "N")
    )
)

skills_35_clean = skills_35_pivot.merge(flags, on=["O*NET-SOC Code", "Element ID"], how="left")

# 5. Validate final shape, duplicates, and missing values
print("Pivoted 35-skills shape (should be 910 x 35 = 31,850 rows):", skills_35_clean.shape)
print("Duplicate (Occupation, Skill) pairs:", skills_35_clean.duplicated(subset=["O*NET-SOC Code", "Element ID"]).sum())
print("\nMissing values:\n", skills_35_clean.isnull().sum())
print("\nFirst 5 rows:")
display(skills_35_clean.head())
```

```python
# 1. Verify join key uniqueness in occupation_raw
print("Duplicate O*NET-SOC Codes in occupation_raw:", occupation_raw["O*NET-SOC Code"].duplicated().sum())

# 2. Check if all 910 occupations in skills_35_clean exist in occupation_raw
skills_codes = set(skills_35_clean["O*NET-SOC Code"].unique())
occ_codes = set(occupation_raw["O*NET-SOC Code"].unique())
missing_in_occ = skills_codes - occ_codes
print("Occupations in skills_35_clean NOT found in occupation_raw:", len(missing_in_occ))

# 3. Merge Description onto skills_35_clean (many-to-one validation prevents row explosion)
career_skills_35 = skills_35_clean.merge(
    occupation_raw[["O*NET-SOC Code", "Description"]],
    on="O*NET-SOC Code",
    how="left",
    validate="many_to_one"
)

# 4. Validate the merged dataset
print("\nMerged dataset shape (should be 31,850 rows, 10 columns):", career_skills_35.shape)
print("\nMissing values after merge:\n", career_skills_35.isnull().sum())
print("\nFirst 3 rows:")
display(career_skills_35.head(3))

# 5. Save a clean checkpoint CSV in Colab
career_skills_35.to_csv("HER360_Career_Skills35_Processed.csv", index=False)
print("\nSaved checkpoint: HER360_Career_Skills35_Processed.csv")
```

```python
# 1. Filter software_skills.csv to only our 910 occupations (keep raw data untouched)
valid_codes = set(career_skills_35["O*NET-SOC Code"].unique())
sw_910 = skill_files["software_skills.csv"].copy()
sw_910 = sw_910[sw_910["O*NET-SOC Code"].isin(valid_codes)].reset_index(drop=True)

print("Filtered software_skills shape:", sw_910.shape)
print("Unique occupations covered:", sw_910["O*NET-SOC Code"].nunique())
print("Duplicate (O*NET-SOC Code, Workplace Example) pairs:",
      sw_910.duplicated(subset=["O*NET-SOC Code", "Workplace Example"]).sum())

# 2. Check distribution of software tools per occupation
tools_per_occ = sw_910.groupby("O*NET-SOC Code")["Workplace Example"].nunique()
print("\nSoftware tools per occupation (summary):")
print(tools_per_occ.describe())

# 3. Preview one concrete occupation (e.g., Chief Executives: 11-1011.00)
sample_code = "11-1011.00"
print("\n" + "=" * 60)
print(f"SAMPLE OCCUPATION: {sample_code}")

top_core = (
    career_skills_35[
        (career_skills_35["O*NET-SOC Code"] == sample_code) &
        (career_skills_35["Not_Relevant"] == "N")
    ]
    .sort_values(by="Importance", ascending=False)
    .head(8)[["Element Name", "Skill_Category", "Importance", "Level"]]
)
print("\nTop 8 Core Skills (by Importance):")
display(top_core)

sample_sw = sw_910[sw_910["O*NET-SOC Code"] == sample_code][
    ["Workplace Example", "Element Name", "Hot Technology", "In Demand"]
].head(10)
print("\nFirst 10 Software Tools:")
display(sample_sw)
```

```python
# 1. Start with the 910 unique occupations from occupation_raw
valid_codes = set(career_skills_35["O*NET-SOC Code"].unique())
occ_master = (
    occupation_raw[occupation_raw["O*NET-SOC Code"].isin(valid_codes)]
    .copy()
    .sort_values("O*NET-SOC Code")
    .reset_index(drop=True)
)

# 2. Summarize top 10 relevant core skills per occupation (sorted by Importance)
relevant_skills = career_skills_35[career_skills_35["Not_Relevant"] == "N"].copy()
relevant_skills = relevant_skills.sort_values(
    ["O*NET-SOC Code", "Importance"], ascending=[True, False]
)

core_summary = (
    relevant_skills.groupby("O*NET-SOC Code")
    .agg(
        Top_10_Core_Skills=("Element Name", lambda s: ", ".join(s.head(10))),
        Relevant_Core_Skill_Count=("Element Name", "count")
    )
    .reset_index()
)

# 3. Summarize software categories, specific tools, and hot technologies per occupation
sw_summary = (
    sw_910.groupby("O*NET-SOC Code")
    .agg(
        Software_Categories=("Element Name", lambda s: ", ".join(sorted(s.unique()))),
        Software_Tools=("Workplace Example", lambda s: ", ".join(sorted(s.unique()))),
        Num_Software_Tools=("Workplace Example", "nunique")
    )
    .reset_index()
)

hot_tech_summary = (
    sw_910[sw_910["Hot Technology"] == "Y"]
    .groupby("O*NET-SOC Code")["Workplace Example"]
    .apply(lambda s: ", ".join(sorted(s.unique())))
    .reset_index(name="Hot_Tech_Tools")
)

# 4. Merge all summaries into the 910-row master table
occ_master = (
    occ_master
    .merge(core_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
    .merge(sw_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
    .merge(hot_tech_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
)

# Some occupations may have 0 "Hot Technology" tools; replace NaN with empty string
occ_master["Hot_Tech_Tools"] = occ_master["Hot_Tech_Tools"].fillna("")

# 5. Validate shape, duplicates, and missing values
print("Master occupation table shape (should be 910 rows):", occ_master.shape)
print("Duplicate occupations:", occ_master["O*NET-SOC Code"].duplicated().sum())
print("\nMissing values:\n", occ_master.isnull().sum())
print("\nFirst 2 rows:")
display(occ_master.head(2))

# 6. Save checkpoint CSV
occ_master.to_csv("HER360_Occupations_Master_910.csv", index=False)
print("\nSaved checkpoint: HER360_Occupations_Master_910.csv")
```

```python
valid_codes = set(occ_master["O*NET-SOC Code"].unique())

# 1. Summarize alternate job titles (capped at 10 per occupation to avoid text noise)
jt_910 = extra_files["job_titles (1).csv"].copy()
jt_910 = jt_910[jt_910["O*NET-SOC Code"].isin(valid_codes)]

jt_summary = (
    jt_910.groupby("O*NET-SOC Code")
    .agg(
        Alternate_Titles_Top10=("Job Title", lambda s: ", ".join(pd.unique(s)[:10])),
        Total_Alternate_Titles=("Job Title", "nunique")
    )
    .reset_index()
)

# 2. Create finalized 910-occupation master table
occ_final = occ_master.copy().merge(
    jt_summary, on="O*NET-SOC Code", how="left", validate="one_to_one"
)

# 3. Build a clean, structured text profile for each occupation (for TF-IDF, BM25, Sentence Transformers)
occ_final["Occupation_Text"] = (
    "Title: " + occ_final["Title"] + ". "
    + "Alternate Titles: " + occ_final["Alternate_Titles_Top10"] + ". "
    + "Description: " + occ_final["Description"] + " "
    + "Core Skills: " + occ_final["Top_10_Core_Skills"] + ". "
    + "Software Categories: " + occ_final["Software_Categories"] + ". "
    + "Hot Technologies: " + occ_final["Hot_Tech_Tools"]
)

# 4. Create a clean related-occupations table (only links where BOTH sides are in our 910 jobs)
rel_final = extra_files["related_occupations.csv"].copy()
rel_final = rel_final[
    rel_final["O*NET-SOC Code"].isin(valid_codes) &
    rel_final["Related O*NET-SOC Code"].isin(valid_codes)
].reset_index(drop=True)

# 5. Validate all finalized tables
print("occ_final shape (should be 910 rows):", occ_final.shape)
print("occ_final missing values:", occ_final.isnull().sum().sum())
print("rel_final shape:", rel_final.shape)
print("sw_910 shape:", sw_910.shape)
print("career_skills_35 shape:", career_skills_35.shape)

# 6. Preview one Occupation_Text example to verify it reads cleanly
print("\n" + "=" * 60)
print("SAMPLE Occupation_Text (Row 0):")
print(occ_final.loc[0, "Occupation_Text"][:600], "...")

# 7. Save finalized O*NET checkpoint files
occ_final.to_csv("HER360_Occupations_Final_910.csv", index=False)
rel_final.to_csv("HER360_Related_Occupations_910.csv", index=False)
sw_910.to_csv("HER360_Software_Skills_910.csv", index=False)
print("\nSaved final O*NET CSVs successfully.")
```
