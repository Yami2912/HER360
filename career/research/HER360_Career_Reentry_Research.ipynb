```python
import pandas as pd

# 1. Copy raw DataFrames so the originals in skill_files stay untouched
ess_copy = skill_files["essential_skills.csv"].copy()
ess_copy["Skill_Category"] = "Essential"

trans_copy = skill_files["transferable_skills.csv"].copy()
trans_copy["Skill_Category"] = "Transferable"

# 2. Stack them vertically into one raw combined skills table
all_skills_raw = pd.concat([ess_copy, trans_copy], ignore_index=True)
print("Combined raw shape (should be 63,700 rows):", all_skills_raw.shape)

# 3. Pivot IM (Importance) and LV (Level) into separate columns
skills_35_pivot = all_skills_raw.pivot(
    index=["O*NET-SOC Code", "Title", "Element ID", "Element Name", "Skill_Category"],
    columns="Scale ID",
    values="Data Value"
).reset_index()

skills_35_pivot = skills_35_pivot.rename(columns={"IM": "Importance", "LV": "Level"})

# 4. Also bring in the quality flags (Not Relevant from LV rows; Recommend Suppress if Y on either IM or LV)
flags = (
    all_skills_raw.groupby(["O*NET-SOC Code", "Element ID"], as_index=False)
    .agg(
        Recommend_Suppress=("Recommend Suppress", lambda s: "Y" if (s == "Y").any() else "N"),
        Not_Relevant=("Not Relevant", lambda s: "Y" if (s == "Y").any() else "N")
    )
)

skills_35_clean = skills_35_pivot.merge(flags, on=["O*NET-SOC Code", "Element ID"], how="left")

# 5. Validate final shape, duplicates, and missing values
print("Pivoted 35-skills shape (should be 910 x 35 = 31,850 rows):", skills_35_clean.shape)
print("Duplicate (Occupation, Skill) pairs:", skills_35_clean.duplicated(subset=["O*NET-SOC Code", "Element ID"]).sum())
print("\nMissing values:\n", skills_35_clean.isnull().sum())
print("\nFirst 5 rows:")
display(skills_35_clean.head())
```

```python
# 1. Start with the 910 unique occupations from occupation_raw
valid_codes = set(career_skills_35["O*NET-SOC Code"].unique())
occ_master = (
    occupation_raw[occupation_raw["O*NET-SOC Code"].isin(valid_codes)]
    .copy()
    .sort_values("O*NET-SOC Code")
    .reset_index(drop=True)
)

# 2. Summarize top 10 relevant core skills per occupation (sorted by Importance)
relevant_skills = career_skills_35[career_skills_35["Not_Relevant"] == "N"].copy()
relevant_skills = relevant_skills.sort_values(
    ["O*NET-SOC Code", "Importance"], ascending=[True, False]
)

core_summary = (
    relevant_skills.groupby("O*NET-SOC Code")
    .agg(
        Top_10_Core_Skills=("Element Name", lambda s: ", ".join(s.head(10))),
        Relevant_Core_Skill_Count=("Element Name", "count")
    )
    .reset_index()
)

# 3. Summarize software categories, specific tools, and hot technologies per occupation
sw_summary = (
    sw_910.groupby("O*NET-SOC Code")
    .agg(
        Software_Categories=("Element Name", lambda s: ", ".join(sorted(s.unique()))),
        Software_Tools=("Workplace Example", lambda s: ", ".join(sorted(s.unique()))),
        Num_Software_Tools=("Workplace Example", "nunique")
    )
    .reset_index()
)

hot_tech_summary = (
    sw_910[sw_910["Hot Technology"] == "Y"]
    .groupby("O*NET-SOC Code")["Workplace Example"]
    .apply(lambda s: ", ".join(sorted(s.unique())))
    .reset_index(name="Hot_Tech_Tools")
)

# 4. Merge all summaries into the 910-row master table
occ_master = (
    occ_master
    .merge(core_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
    .merge(sw_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
    .merge(hot_tech_summary, on="O*NET-SOC Code", how="left", validate="one_to_one")
)

# Some occupations may have 0 "Hot Technology" tools; replace NaN with empty string
occ_master["Hot_Tech_Tools"] = occ_master["Hot_Tech_Tools"].fillna("")

# 5. Validate shape, duplicates, and missing values
print("Master occupation table shape (should be 910 rows):", occ_master.shape)
print("Duplicate occupations:", occ_master["O*NET-SOC Code"].duplicated().sum())
print("\nMissing values:\n", occ_master.isnull().sum())
print("\nFirst 2 rows:")
display(occ_master.head(2))

# 6. Save checkpoint CSV
occ_master.to_csv("HER360_Occupations_Master_910.csv", index=False)
print("\nSaved checkpoint: HER360_Occupations_Master_910.csv")
```

```python
import numpy as np
import pandas as pd

# 1. Individual ranking metrics (for one user profile)
def precision_at_k(recommended, relevant_set, k=5):
    top_k = recommended[:k]
    hits = sum(1 for item in top_k if item in relevant_set)
    return hits / k

def recall_at_k(recommended, relevant_set, k=5):
    if not relevant_set:
        return 0.0
    top_k = recommended[:k]
    hits = sum(1 for item in top_k if item in relevant_set)
    return hits / len(relevant_set)

def ndcg_at_k(recommended, relevant_set, k=5):
    top_k = recommended[:k]
    dcg = sum(1.0 / np.log2(rank + 2) for rank, item in enumerate(top_k) if item in relevant_set)
    ideal_hits = min(len(relevant_set), k)
    idcg = sum(1.0 / np.log2(rank + 2) for rank in range(ideal_hits))
    return dcg / idcg if idcg > 0 else 0.0

def average_precision_at_k(recommended, relevant_set, k=5):
    if not relevant_set:
        return 0.0
    top_k = recommended[:k]
    hits = 0
    score = 0.0
    for rank, item in enumerate(top_k):
        if item in relevant_set:
            hits += 1
            score += hits / (rank + 1)
    return score / min(len(relevant_set), k)

# 2. Shared evaluation runner (works for ANY recommender model)
def evaluate_model(model_name, rank_fn, profiles, k=5, catalog_size=910):
    p_scores, r_scores, ndcg_scores, ap_scores = [], [], [], []
    all_recommended_top_k = set()

    for profile in profiles:
        rel_set = set(profile["relevant_occupations"])
        ranked_codes = rank_fn(profile)

        p_scores.append(precision_at_k(ranked_codes, rel_set, k))
        r_scores.append(recall_at_k(ranked_codes, rel_set, k))
        ndcg_scores.append(ndcg_at_k(ranked_codes, rel_set, k))
        ap_scores.append(average_precision_at_k(ranked_codes, rel_set, k))
        all_recommended_top_k.update(ranked_codes[:k])

    return {
        "Model": model_name,
        f"Precision@{k}": round(float(np.mean(p_scores)), 4),
        f"Recall@{k}": round(float(np.mean(r_scores)), 4),
        f"nDCG@{k}": round(float(np.mean(ndcg_scores)), 4),
        f"MAP@{k}": round(float(np.mean(ap_scores)), 4),
        f"Unique_Jobs@{k}": len(all_recommended_top_k),
        f"Coverage@{k}_%": round((len(all_recommended_top_k) / catalog_size) * 100, 2)
    }

# 3. Split benchmark cleanly into Validation and Test lists
val_profiles = [p for p in eval_profiles if p["split"] == "val"]
test_profiles = [p for p in eval_profiles if p["split"] == "test"]

# 4. Sanity check: Dummy Perfect vs Dummy Wrong recommender on val_profiles
dummy_perfect = lambda p: p["relevant_occupations"] + ["00-0000.00"] * 5
dummy_wrong = lambda p: ["00-0000.00"] * 5

sanity_results = pd.DataFrame([
    evaluate_model("Sanity Check - Perfect", dummy_perfect, val_profiles, k=5),
    evaluate_model("Sanity Check - All Wrong", dummy_wrong, val_profiles, k=5)
])
display(sanity_results)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

# 1. Shared helper: convert a user profile into one clean text query
# (Used by TF-IDF, BM25, and Sentence Transformers so all text models get the same input)
def profile_to_text(profile):
    skills_str = ", ".join(profile["core_skills"])
    tools_str = ", ".join(profile["software_tools"])
    return f"{profile['summary_text']} Skills: {skills_str}. Tools: {tools_str}."

# 2. Fit TF-IDF on all 910 occupation text profiles
occ_codes_ordered = occ_final["O*NET-SOC Code"].values
occ_titles_map = dict(zip(occ_final["O*NET-SOC Code"], occ_final["Title"]))

tfidf_vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),   # captures single words ("Python") and 2-word phrases ("data analysis")
    min_df=1
)
occ_tfidf_matrix = tfidf_vectorizer.fit_transform(occ_final["Occupation_Text"])

# 3. Scoring and ranking functions for TF-IDF
def get_tfidf_scores(profile):
    query_vec = tfidf_vectorizer.transform([profile_to_text(profile)])
    scores = cosine_similarity(query_vec, occ_tfidf_matrix).flatten()
    return scores

def rank_tfidf(profile):
    scores = get_tfidf_scores(profile)
    ranked_idx = np.argsort(scores)[::-1]
    return occ_codes_ordered[ranked_idx].tolist()

# 4. Evaluate on Validation Set ONLY (keep Test Set untouched for now)
val_results = []
tfidf_val_metrics = evaluate_model("1. TF-IDF + Cosine", rank_tfidf, val_profiles, k=5)
val_results.append(tfidf_val_metrics)

print("Validation Results (so far):")
display(pd.DataFrame(val_results))

# 5. Inspect actual Top-5 recommendations for the first 3 validation profiles
print("\nSample Top-5 TF-IDF Recommendations (Validation Set):")
for p in val_profiles[:3]:
    top5 = rank_tfidf(p)[:5]
    rel_set = set(p["relevant_occupations"])
    print("=" * 70)
    print(f"Profile: {p['profile_id']} | True Relevant: {[occ_titles_map[c] for c in rel_set]}")
    for rank, code in enumerate(top5, start=1):
        hit = "HIT" if code in rel_set else "   "
        print(f"  #{rank} [{hit}] {code} - {occ_titles_map[code]}")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# 1. Build a (910 x 35) Importance*Level matrix aligned with occ_codes_ordered
skills_mat_df = career_skills_35.copy()
skills_mat_df["Skill_Weight"] = np.where(
    skills_mat_df["Not_Relevant"] == "Y",
    0.0,
    (skills_mat_df["Importance"] / 5.0) * (skills_mat_df["Level"] / 7.0)
)

occ_skill_pivot = (
    skills_mat_df.pivot(index="O*NET-SOC Code", columns="Element Name", values="Skill_Weight")
    .reindex(occ_codes_ordered)
    .fillna(0.0)
)

skill_columns = occ_skill_pivot.columns.tolist()
skill_to_idx = {name: i for i, name in enumerate(skill_columns)}
occ_skill_matrix = occ_skill_pivot.values  # shape: (910, 35)

# 2. Pre-group software tools per occupation for fast matching
occ_tools_map = {code: [] for code in occ_codes_ordered}
for row in sw_910[["O*NET-SOC Code", "Workplace Example", "Hot Technology"]].itertuples(index=False):
    occ_tools_map[row[0]].append((str(row[1]).lower(), row[2] == "Y"))

# 3. Weighted Skill + Software scoring function
def get_skill_scores(profile):
    # (a) Core 35-skill weighted cosine similarity
    user_vec = np.zeros((1, len(skill_columns)), dtype=float)
    for skill in profile["core_skills"]:
        if skill in skill_to_idx:
            user_vec[0, skill_to_idx[skill]] = 1.0

    if user_vec.sum() > 0:
        core_scores = cosine_similarity(user_vec, occ_skill_matrix).flatten()
    else:
        core_scores = np.zeros(len(occ_codes_ordered), dtype=float)

    # (b) Software tool match score (word-boundary safe)
    user_tools = [t.strip().lower() for t in profile["software_tools"] if t.strip()]
    sw_scores = np.zeros(len(occ_codes_ordered), dtype=float)

    if user_tools:
        patterns = [re.compile(r"(?<!\w)" + re.escape(t) + r"(?!\w)") for t in user_tools]
        max_possible = len(user_tools) * 1.2

        for idx, code in enumerate(occ_codes_ordered):
            occ_tools = occ_tools_map[code]
            matched_weight = 0.0
            for pat in patterns:
                # Find if this user tool matches any tool in the occupation
                best_match = 0.0
                for tool_text, is_hot in occ_tools:
                    if pat.search(tool_text):
                        best_match = max(best_match, 1.2 if is_hot else 1.0)
                matched_weight += best_match
            sw_scores[idx] = matched_weight / max_possible

    # Combine Core Skills (50%) + Software Tools (50%)
    return 0.5 * core_scores + 0.5 * sw_scores

def rank_skill(profile):
    scores = get_skill_scores(profile)
    ranked_idx = np.argsort(scores)[::-1]
    return occ_codes_ordered[ranked_idx].tolist()

# 4. Evaluate on Validation Set ONLY and update comparison table
skill_val_metrics = evaluate_model("4. Weighted Skill Matching", rank_skill, val_profiles, k=5)
val_results = [r for r in val_results if r["Model"] != "4. Weighted Skill Matching"]
val_results.append(skill_val_metrics)

print("Validation Comparison (4 Individual Candidate Models):")
display(pd.DataFrame(val_results))

# 5. Inspect Top-5 Weighted Skill recommendations for the first 3 validation profiles
print("\nSample Top-5 Weighted Skill Recommendations (Validation Set):")
for p in val_profiles[:3]:
    top5 = rank_skill(p)[:5]
    rel_set = set(p["relevant_occupations"])
    print("=" * 70)
    print(f"Profile: {p['profile_id']} | True Relevant: {[occ_titles_map[c] for c in rel_set]}")
    for rank, code in enumerate(top5, start=1):
        hit = "HIT" if code in rel_set else "   "
        print(f"  #{rank} [{hit}] {code} - {occ_titles_map[code]}")
```

```python
import numpy as np
import pandas as pd

# 1. Min-Max normalization helper (scales any 910-length score array to [0, 1])
def minmax_norm(scores):
    s_min, s_max = np.min(scores), np.max(scores)
    if s_max - s_min < 1e-12:
        return np.zeros_like(scores, dtype=float)
    return (scores - s_min) / (s_max - s_min)

# 2. Explicit preference score (1.0 if occupation matches user's chosen major group, else 0.0)
occ_major_groups = np.array([code[:2] for code in occ_codes_ordered])

def get_preference_scores(profile):
    prefs = set(profile.get("preferred_major_groups", []))
    if not prefs:
        return np.zeros(len(occ_codes_ordered), dtype=float)
    return np.array([1.0 if g in prefs else 0.0 for g in occ_major_groups], dtype=float)

# 3. Precompute normalized component scores for Validation profiles (makes grid search instant)
val_components = {}
for p in val_profiles:
    pid = p["profile_id"]
    val_components[pid] = {
        "sem": minmax_norm(get_semantic_scores(p)),
        "skill": minmax_norm(get_skill_scores(p)),
        "kw": minmax_norm(get_tfidf_scores(p)),
        "pref": get_preference_scores(p)
    }

# Helper to evaluate any (alpha, beta, gamma, delta) weight combination on val_profiles
def eval_hybrid_weights(alpha, beta, gamma, delta, profiles, comp_cache, k=5):
    def rank_fn(p):
        c = comp_cache[p["profile_id"]]
        score = alpha * c["sem"] + beta * c["skill"] + gamma * c["kw"] + delta * c["pref"]
        return occ_codes_ordered[np.argsort(score)[::-1]].tolist()
    return evaluate_model("temp", rank_fn, profiles, k=k)

# 4. Grid search over weights (step 0.1, summing to 1.0) on VALIDATION SET ONLY
best_no_pref = None
best_with_pref = None

for a_int in range(0, 11):
    for b_int in range(0, 11 - a_int):
        for g_int in range(0, 11 - a_int - b_int):
            d_int = 10 - a_int - b_int - g_int
            a, b, g, d = a_int / 10, b_int / 10, g_int / 10, d_int / 10

            res = eval_hybrid_weights(a, b, g, d, val_profiles, val_components, k=5)
            # Rank combinations by nDCG@5, then MAP@5, then Recall@5
            key = (res["nDCG@5"], res["MAP@5"], res["Recall@5"])

            if d_int == 0:
                if best_no_pref is None or key > best_no_pref[0]:
                    best_no_pref = (key, (a, b, g, d), res)

            if best_with_pref is None or key > best_with_pref[0]:
                best_with_pref = (key, (a, b, g, d), res)

# 5. Store the best learned weights for later Test Set evaluation
best_weights_no_pref = best_no_pref[1]
best_weights_pref = best_with_pref[1]

print(f"Best Hybrid (No Preference) weights   [alpha, beta, gamma, delta]: {best_weights_no_pref}")
print(f"Best Preference-Aware Hybrid weights  [alpha, beta, gamma, delta]: {best_weights_pref}")

# 6. Update Validation Comparison Table
row_no_pref = dict(best_no_pref[2])
row_no_pref["Model"] = f"5a. Hybrid (No Pref) {best_weights_no_pref}"

row_with_pref = dict(best_with_pref[2])
row_with_pref["Model"] = f"5b. Hybrid (Pref-Aware) {best_weights_pref}"

val_results = [r for r in val_results if not r["Model"].startswith("5")]
val_results.extend([row_no_pref, row_with_pref])

print("\nFull Validation Comparison (All Candidate Models):")
display(pd.DataFrame(val_results))
```

```python
import re
import numpy as np
import pandas as pd

# 1. Pre-index top required core skills per occupation (Importance >= 3.0, not marked Not_Relevant)
req_skills_df = career_skills_35[
    (career_skills_35["Not_Relevant"] != "Y") &
    (career_skills_35["Importance"] >= 3.0)
].copy()
req_skills_df["Priority"] = (req_skills_df["Importance"] / 5.0) * (req_skills_df["Level"] / 7.0)
req_skills_df = req_skills_df.sort_values(["O*NET-SOC Code", "Priority"], ascending=[True, False])

occ_req_skills_map = (
    req_skills_df.groupby("O*NET-SOC Code")["Element Name"]
    .apply(list)
    .to_dict()
)

# 2. Pre-index related occupations (prioritizing Primary-Short and Primary-Long)
tier_order = {"Primary-Short": 0, "Primary-Long": 1, "Supplemental": 2}
rel_sorted = rel_final.copy()
rel_sorted["Tier_Rank"] = rel_sorted["Relatedness Tier"].map(tier_order).fillna(3)
rel_sorted = rel_sorted.sort_values(["O*NET-SOC Code", "Tier_Rank"])

occ_related_map = (
    rel_sorted.groupby("O*NET-SOC Code")["Related O*NET-SOC Code"]
    .apply(lambda s: [occ_titles_map[c] for c in s[:3] if c in occ_titles_map])
    .to_dict()
)

# 3. End-to-end recommender + skill-gap analyzer using our locked-in Hybrid weights
LOCKED_WEIGHTS = (0.5, 0.2, 0.2, 0.1)  # (alpha_sem, beta_skill, gamma_kw, delta_pref)

def recommend_with_skill_gap(profile, top_k=3):
    a, b, g, d = LOCKED_WEIGHTS

    sem = minmax_norm(get_semantic_scores(profile))
    skill = minmax_norm(get_skill_scores(profile))
    kw = minmax_norm(get_tfidf_scores(profile))
    pref = get_preference_scores(profile)

    final_scores = a * sem + b * skill + g * kw + d * pref
    top_indices = np.argsort(final_scores)[::-1][:top_k]

    user_skills_set = set(profile.get("core_skills", []))
    user_tools = [t.strip().lower() for t in profile.get("software_tools", []) if t.strip()]
    user_tool_patterns = [re.compile(r"(?<!\w)" + re.escape(t) + r"(?!\w)") for t in user_tools]

    recommendations = []
    for idx in top_indices:
        code = occ_codes_ordered[idx]
        title = occ_titles_map[code]

        # Core skill matches & gaps (top 8 required skills for this job)
        req_skills = occ_req_skills_map.get(code, [])[:8]
        matched_skills = [s for s in req_skills if s in user_skills_set]
        missing_skills = [s for s in req_skills if s not in user_skills_set][:5]

        # Software tool matches & missing Hot Technologies
        occ_tools = occ_tools_map.get(code, [])
        matched_tools = []
        missing_hot_tools = []

        for tool_text, is_hot in occ_tools:
            if any(pat.search(tool_text) for pat in user_tool_patterns):
                if tool_text not in matched_tools:
                    matched_tools.append(tool_text)
            elif is_hot and tool_text not in missing_hot_tools:
                missing_hot_tools.append(tool_text)

        recommendations.append({
            "O*NET-SOC Code": code,
            "Title": title,
            "Hybrid_Score": round(float(final_scores[idx]), 4),
            "Matched_Core_Skills": matched_skills,
            "Priority_Skill_Gaps": missing_skills,
            "Matched_Tools": matched_tools[:5],
            "Top_Hot_Tech_to_Learn": missing_hot_tools[:5],
            "Related_Transitions": occ_related_map.get(code, [])
        })

    return recommendations

# 4. Preview the full output on 2 Test profiles (Tech/Data + Counseling)
for sample_profile in test_profiles:
    print("=" * 80)
    print(f"PROFILE: {sample_profile['profile_id']} | {sample_profile['summary_text']}")
    recs = recommend_with_skill_gap(sample_profile, top_k=3)
    for rank, r in enumerate(recs, start=1):
        print(f"\n  #{rank} {r['Title']} ({r['O*NET-SOC Code']})  [Match Score: {r['Hybrid_Score']}]")
        print(f"     Matched Skills:       {r['Matched_Core_Skills']}")
        print(f"     Priority Skill Gaps:  {r['Priority_Skill_Gaps']}")
        print(f"     Matched Tools:        {r['Matched_Tools']}")
        print(f"     Hot Tech to Learn:    {r['Top_Hot_Tech_to_Learn']}")
        print(f"     Related Careers:      {r['Related_Transitions']}")
```
