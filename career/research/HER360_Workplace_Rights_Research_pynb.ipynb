```python
!pip install -q rank_bm25 sentence-transformers

import re
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer

doc_ids = rights_corpus_df["doc_id"].values
corpus_texts = rights_corpus_df["search_text"].tolist()

# 1. Model A: TF-IDF + Cosine
tfidf_vec = TfidfVectorizer(stop_words="english", ngram_range=(1, 2))
corpus_tfidf = tfidf_vec.fit_transform(corpus_texts)

def score_tfidf(query):
    return cosine_similarity(tfidf_vec.transform([query]), corpus_tfidf).flatten()

# 2. Model B: BM25
def tokenize(text):
    return re.findall(r"[a-z0-9]+", str(text).lower())

bm25 = BM25Okapi([tokenize(doc) for doc in corpus_texts])

def score_bm25(query):
    return np.array(bm25.get_scores(tokenize(query)), dtype=float)

# 3. Model C: Dense Embeddings (all-MiniLM-L6-v2)
embed_model = SentenceTransformer("all-MiniLM-L6-v2")
corpus_emb = embed_model.encode(corpus_texts, normalize_embeddings=True)

def score_dense(query):
    q_emb = embed_model.encode([query], normalize_embeddings=True)[0]
    return np.array(np.dot(corpus_emb, q_emb), dtype=float)

# 4. Evaluation Helper (In-Scope Queries Only)
in_scope_queries = [q for q in EVAL_QUERIES if q["is_answerable"]]
oos_queries = [q for q in EVAL_QUERIES if not q["is_answerable"]]

def evaluate_retriever(name, score_fn, k=3):
    recalls, precisions, mrrs = [], [], []
    for q in in_scope_queries:
        scores = score_fn(q["query"])
        top_k_docs = doc_ids[np.argsort(scores)[::-1][:k]].tolist()
        rel = set(q["expected_docs"])

        hits = sum(1 for d in top_k_docs if d in rel)
        recalls.append(hits / len(rel))
        precisions.append(hits / k)

        rr = 0.0
        for rank, d in enumerate(top_k_docs, start=1):
            if d in rel:
                rr = 1.0 / rank
                break
        mrrs.append(rr)

    return {
        "Retriever": name,
        "Recall@3": round(float(np.mean(recalls)), 4),
        "Precision@3": round(float(np.mean(precisions)), 4),
        "MRR": round(float(np.mean(mrrs)), 4)
    }

retrieval_results = pd.DataFrame([
    evaluate_retriever("1. TF-IDF + Cosine", score_tfidf, k=3),
    evaluate_retriever("2. BM25", score_bm25, k=3),
    evaluate_retriever("3. Dense (all-MiniLM-L6-v2)", score_dense, k=3),
])

print("FEATURE B RETRIEVAL COMPARISON (12 In-Scope Queries, K=3):")
display(retrieval_results)

# 5. Check Dense Similarity Score Separation (In-Scope vs Out-of-Scope) for Abstention Threshold
in_scope_max = [float(np.max(score_dense(q["query"]))) for q in in_scope_queries]
oos_max = [float(np.max(score_dense(q["query"]))) for q in oos_queries]

print("\nDense Top-1 Similarity Score Distribution (for Abstention Calibration):")
print(f"  In-Scope (12 queries):      min={min(in_scope_max):.4f}, max={max(in_scope_max):.4f}, mean={np.mean(in_scope_max):.4f}")
print(f"  Out-of-Scope (4 queries):   min={min(oos_max):.4f}, max={max(oos_max):.4f}, mean={np.mean(oos_max):.4f}")
```

```python
import json
import numpy as np
import pandas as pd

ABSTENTION_THRESHOLD = 0.44
DISCLAIMER = (
    "Disclaimer: This response is for general workplace-rights awareness in India and does not constitute "
    "legal advice. HER360 is not an official complaint portal. For formal action or case-specific guidance, "
    "consult a qualified legal professional or use official Government of India channels."
)

def ask_workplace_rights_assistant(query, top_k=2, threshold=ABSTENTION_THRESHOLD):
    scores = score_dense(query)
    top_idx = np.argsort(scores)[::-1][:top_k]
    top_score = float(scores[top_idx[0]])

    # Safe Abstention Gate
    if top_score < threshold:
        return {
            "query": query,
            "status": "ABSTAINED",
            "confidence_score": round(top_score, 4),
            "answer": (
                "I cannot provide a reliable, source-grounded answer to this question from my verified "
                "Indian workplace-rights corpus. Please consult a qualified legal professional or check "
                "official Government of India portals (India Code: https://www.indiacode.nic.in/, "
                "Ministry of Labour & Employment: https://labour.gov.in/)."
            ),
            "citations": [],
            "disclaimer": DISCLAIMER
        }

    # Grounded Answer strictly from retrieved provisions
    citations = []
    answer_parts = []
    for idx in top_idx:
        if float(scores[idx]) >= threshold - 0.05:
            row = rights_corpus_df.iloc[idx]
            answer_parts.append(f"[{row['section']} - {row['topic']}]: {row['text']}")
            citations.append({
                "doc_id": row["doc_id"],
                "act_name": row["act_name"],
                "section": row["section"],
                "official_channel": row["official_channel"],
                "source_url": row["source_url"],
                "last_verified_date": row["last_verified_date"],
                "similarity_score": round(float(scores[idx]), 4)
            })

    return {
        "query": query,
        "status": "ANSWERED",
        "confidence_score": round(top_score, 4),
        "answer": " ".join(answer_parts),
        "citations": citations,
        "disclaimer": DISCLAIMER
    }

# 1. Evaluate Abstention & Grounding across all 16 queries
eval_logs = []
for item in EVAL_QUERIES:
    resp = ask_workplace_rights_assistant(item["query"])
    predicted_answerable = (resp["status"] == "ANSWERED")
    correct_decision = (predicted_answerable == item["is_answerable"])
    top_doc = resp["citations"][0]["doc_id"] if resp["citations"] else "ABSTAINED"

    eval_logs.append({
        "Query_ID": item["query_id"],
        "Expected_Answerable": item["is_answerable"],
        "Assistant_Status": resp["status"],
        "Top_Score": resp["confidence_score"],
        "Top_Cited_Doc": top_doc,
        "Decision_Correct": correct_decision
    })

eval_logs_df = pd.DataFrame(eval_logs)
in_scope_acc = eval_logs_df[eval_logs_df["Expected_Answerable"]]["Decision_Correct"].mean()
oos_acc = eval_logs_df[~eval_logs_df["Expected_Answerable"]]["Decision_Correct"].mean()

print(f"In-Scope Answer Rate:          {in_scope_acc:.2%} (threshold={ABSTENTION_THRESHOLD})")
print(f"Out-of-Scope Abstention Rate:  {oos_acc:.2%}")
display(eval_logs_df)

# 2. Preview & Save 2 sample outputs (1 In-Scope POSH/SHe-Box + 1 Out-of-Scope Abstention)
sample_outputs = [
    ask_workplace_rights_assistant("Is there an official Indian government website where I can file a workplace sexual harassment complaint online?"),
    ask_workplace_rights_assistant("How do I claim Section 80C deductions on my Indian income tax return?")
]

with open("HER360_FeatureB_Sample_Output.json", "w") as f:
    json.dump(sample_outputs, f, indent=2)

print("\nSAMPLE GROUNDED OUTPUTS:")
print(json.dumps(sample_outputs, indent=2))
print("\nSaved: HER360_FeatureB_Sample_Output.json")
```
